## Business UnderStanding

Wines are living drinks in constant evolution, needing ideal conditions to keep their personality and to not degrade, requiring special care in its keep **&** transportation in order not to loose inventory, making demand prediction very important as it can make a difference in revenue more than some other areas.

<center>
    <img src=https://images.unsplash.com/photo-1596438214057-5ff7c7fa76b1>
</center>

It is necessary to avoid sunlight and keep a certain level of humidity in the environment, without excess, keeping a constant temperature, which is the most important, more than the actual temperature itself, although it ranges between 10**º**C **&** 15**º**C, depending on the type of wine.

It is also important to avoid movement, shaking, vibrations, or any type of disturbance to the liquid, which means inventory management should be optimized to avoid moving the wines more than necessary. It is also recommended that the bottles be stored horizontally whenever possible, which should be considered when planning the storage area.

For this reason, it is essential to think about optimizing stock rotation, so that wine spends the least amount of time possible in a position that could deteriorate it, ensuring it is sold while still in excellent condition.

Although some wines improve with age, the vast majority are made for **"**immediate**"** consumption, from the moment they reach the market, up to a maximum of 1 to 3 years, making it crucial not to keep wines in stock for too long.

## Time Series

Time series data is a sequence of observations recorded at regular intervals over time. This type of data is ubiquitous, found in fields like finance **(**stock prices**)**, meteorology **(**temperature**)**, and economics **(**GDP**)**, for example. Analyzing time series data helps to understand trends, seasonality, and patterns within the data, enabling informed predictions **&** decisions. Applying statistical techniques **&** machine learning models allows to uncover valuable InSights **&** ForeCast future values, aiding in areas like inventory management, demand forecasting, risk assessment, and more.

By leveraging ProPhet**'**s capabilities, data**-**driven decisions can be made, improving operational efficiency, gaining a competitive edge.

```python
# @title ProPhet
# DataSet Preparation:
clean.reset_index(inplace=True)
ts=clean.groupby('date',as_index=False)['sales'].sum()
ts['date']=pd.to_datetime(ts['date'], format='%Y-%m-%d')
ts.index  =pd.to_datetime(ts['date'], format='%Y-%m-%d')
ts.drop('date', axis=1, inplace=True)
# ReNaming:
ts=ts.reset_index()
ts.columns=['ds','y']
ts#.head()
```

```python
# Extracting Values:
X=ts.y
# Applying ADF:
result=adfuller(X)
print('Augmented Dickey–Fuller:')
print('Statistical Test: {:.4f}'.format( result[0]))
print('P-Value:           {:.4f}'.format(result[1]))
print('Critical Values:')
for key, value in result[4].items(): print('\t{}: {:.4f}'.format(key, value))
```

```python
# Defining Prediction for a 30-Day Period:
prediction_size=30
train=ts[:-prediction_size]
# Instantiating the Model:
b=Prophet()
#      Training the Model:
b.fit(train)
# Making Predictions:
future  =b.make_future_dataframe(periods=prediction_size)
forecast=b.predict(future)
forecast#.head()
```

```python
# Plotting Predictions:
b.plot(forecast).savefig('ForeCastNonStationary.png')
plt.xlabel('')
plt.ylabel('')
plt.box(False)
plt.show(    )
```

```python
# Plotting Components:
b.plot_components(forecast).savefig('NonStationaryComponents.png')
sns.despine(top=True,  left=True,    bottom=True,     right=True )
plt.show(  )
```

```python
# Defining a Function:
def MakeComparisonDataFrame(historical, forecast):
    return forecast.set_index('ds')[['yhat','yhat_lower','yhat_upper']].join(historical.set_index('ds'))
cmp_df=MakeComparisonDataFrame(ts, forecast)
def CalculateForeCastErrors(ts, prediction_size):
    df      = ts.copy()
    df['e'] = df['y'] - df['yhat']
    df['p'] = 100 * df['e'] / df['y']
    predicted_part= df[-prediction_size:]
    error_mean    = lambda error_name: np.mean(np.abs(predicted_part[error_name]))
    return{'MAPE':error_mean('p'),'MAE':error_mean('e')}
# MAPE & MAE:
for err_name, err_value in CalculateForeCastErrors(cmp_df, prediction_size).items(): print(err_name, err_value)
```

```python
# Visualsying Results:
fig, ax=plt.subplots(figsize=(15, 7.5))
ax.plot( cmp_df['yhat_upper'], label='yhat_upper', color='#00BFFF')
ax.plot( cmp_df['yhat'      ], label='yhat'      , color='#00FFFF')
ax.plot( cmp_df['yhat_lower'], label='yhat_lower', color='#6595EE')
ax.plot( cmp_df['y'         ], label='y'         , color='#0065FF')
ax.legend(   )
plt.xlabel('')
plt.ylabel('Average Weekly Sales')
plt.box(False)
# plt.savefig('NonStationaryPredictions.png')
plt.show(    )
```
