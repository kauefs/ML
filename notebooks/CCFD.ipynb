# Credit Card Fraud Detection **(**CCFD**)**

Payment card fraud is a major challenge for business **&** card issuers, as well as transactional services companies, causing substantial **&** growing financial losses. According to the 2019 Nilson Report, card fraud losses worldwide have increased from 9**.**84 billion dollars in 2011 to 27**.**85 billion dollars in 2018, and are projected to reach more than 40 billion dollars in 2027.

Detecting fraud patterns in payment card transactions is a very difficult problem, changing over time. With the ever**-**growing amount of data generated by payment card transactions, it has become impossible for a human analyst to detect fraudulent patterns in transaction datasets, often characterized by a large number of samples, many dimensions, and online updates. As a result, the design of payment card fraud detection techniques has increasingly focused in the last decade on approaches based on machine learning **(**ML**)** techniques **–** algorithms that improve automatically through experience, automating the process of identifying fraudulent patterns from large volumes of data, greatly improving the ability to more efficiently detect frauds **&** identify illicit transactions, a great opportunity for Data Science. In the 1990s, automated decision**-**making changed the banking **&** cosumer credit industries dramatically. Though in recent years the number of fraudulent transactions kept on increasing, the percentage of losses due to fraud started to decrease in 2016, a reverse trend that is associated with the increasing adoption of ML solutions. On top of helping saving money, nowadays, implementing ML**-**based fraud detection systems is becoming a must**-**do for institutions **&** companies to gain customers**'** trust, but as technology evolves, so do fraudster techniques, they adapt, benefiting from the constant changes in volume **&** characteristics of genuine transactions.

The challenge is to improve algorithms in order to minimize or block fraudulent transactions while also keeping false positives very low, that is, those times that a legit purchase is blocked, stressing the consumer.

There are two main scenarios while using a card for transactions**:**
* The physical card is     present **(**CP**):** point**-**of**-**sale **(**POS**)** or automated teller machine **(**ATM**)**
* The physical card is not present **(**CNP**):** usually, online payment.

The first two layers **(**Terminal **&** Transaction Blocking Rules**)** are executed in real**-**time **(**i**.**e**.** within milliseconds **&** before authorization**)**. The next two layers **(**Scoring Rules **&** Data**-**Driven Model**)**, are executed in near real**-**time to potentially block the card and prevent additional frauds. The last layer **(**Investigators**)** is the only one requiring human intervention and is carried out offline. Rules are expressed as _if**-**then**(-**else**)**_ statements.

The Data**-**Driven Model **(**DDM**)** helps investigators by raising alerts on the most suspicious transactions, adopting a classifier or another statistical model to estimate the probability for each feature vector to be a fraud, trained from a set of labeled transactions, simultaneously analyzing multiple components of the feature vector, possibly through nonlinear expressions to find frauds according to rules that go beyond the investigator experience, and that do not necessarily correspond to interpretable rules.

## Data

The [DataSet](https://www.kaggle.com/mlg-ulb/creditcardfraud) contains transactions made by credit cards in September 2013 by European CardHolders, presenting transactions that occurred in two days, where 492 out of 284**,**807 transactions were fraudulent. It is [highly unbalanced](http://sigmoidal.ai/como-lidar-com-dados-desbalanceados/), with the positive class **(**frauds**)** accounting for 0**.**17**%** of all transactions, interferring in the model**'**s performance.

It contains only numerical input variables as a result of a [PCA](http://www2.ic.uff.br/~aconci/PCA-ACP.pdf) **(**Principal Component Analysis**)** transformation. Due to confidentiality issues, original features and more background information about the data cannot be provided. Features $[V1, V2, V3 \dots, V28]$ are the principal components obtained with PCA. The only features which have not been transformed with PCA are `Time` **(**containing the seconds elapsed between each transaction and the first transaction in the dataset**)** **&** `Amount` **(**the transaction Amount, used for example**-**dependant **&** cost**-**sensitive learning**)**. The feature `Class` is the response variable and it takes value 1 in case of fraud and 0 otherwise.

Given the class imbalance ratio, accuracy would be better measured using the Area Under the Precision**-**ReCall Curve **(**AUPRC**)** as Confusion Matrix Accuracy is not meaningful for unbalanced classification.

Fraud detection is a typical example of unbalanced data, where ordinary transactions are way higher than fraudulent ones. Withoutproper balancing, a classifier would return too many false negatives, that is, when a fraudulent transaction is incorrectly identified as legitimate, which would be very inconvenient for the financial institution. A trade**-**-off is to err on false positives, when a genuine transaction is flagged as fraud, placing the inconvenience on customers, who may have the transaction denied, even though they are doing the transaction with theier own card, which, instead, should have been a true negative, a legitimate transaction. The objective being, naturally, correctly identify and block true postive transactions **(**fraudulent activity**)**.

In order to deal with data inbalance, a sampling approach will be used. It is a simple approach but that has been widespread used with good results. In this case, under**-**sampling would be a better approach, randomly eliminating entries from the major class, but preserving the minority class, being ideal for a great amount of data.

```python
# @title Standardization
# Standardization with Standard Scaler since OutLiers are Present:
clean=df.copy()
std  =StandardScaler()
clean[      'amount']=std.fit_transform(clean['Amount'].values.reshape(-1, 1))
clean[      'time'  ]=std.fit_transform(clean['Time'  ].values.reshape(-1, 1))
clean.drop(['Time', 'Amount'], axis=1,    inplace=True)
clean#.head()
```

```python
# @title Train & Validation & Test
# Feature Matrix & Target Vector Split:
XX=clean.drop('Class', axis=1)
YY=        df['Class']
# Train & Validation Split:
X, xx, Y, yy=train_test_split(XX, YY, stratify=YY, shuffle=True)# class proportion for Train & Validation
print('\ntrainX dimensions:' ,  X.shape,
      '\nval  X dimensions: ', xx.shape,
      '\ntrainY dimensions:' ,  Y.shape,
      '\nval  Y dimensions: ', yy.shape)
```

```python
# @title Data Balancing
# UnderSampling Preserving Minoriy Class:
rus=RandomUnderSampler()
rusX, rusY=rus.fit_resample(X, Y)
# Printing Balanced Dimensions:
print(pd.Series(rusY).value_counts())
print('\nrusX dimensions:', rusX.shape,
      '\nrusY dimensions:', rusY.shape)
# Plotting New Class Distribution:
sns.countplot(x=rusY, color='#6595EE'    ,               width=.75   )
sns.despine(     )
plt.title( 'Balanced Class Distribution:', fontdict=FontT, loc='left')
plt.ylabel('Count'                       , fontdict=FontY)
plt.xlabel('Class'                       , fontdict=FontX)
plt.box(  False  )
plt.tight_layout()
plt.show(        )
```

```python
# Balanced Transaction Distribution by Hour: https://www.kaggle.com/code/korcy78/fraud-shield-ai-powered-defense
# Legit & Fraud Split:
legit=df[df['Class']==0]
fraud=df[df['Class']==1]
# Sample the Same Number of Legit Transactions as Fraud Transactions:
balance =legit.sample(len(fraud)*3, random_state=42)
# Combine Fraud & Balanced Legit Data:
balanced=pd.concat([fraud, balance])
# Adding 'Hour' Feature:
balanced['Hour']=(balanced['Time']//3600)%24
# Creating HistoGram:
plt.figure(figsize=(10, 6.5))
sns.histplot(data=balanced, x='Hour', hue='Class', multiple='stack', bins=24, palette=['#6595EE','#00BFFF'])#['#5D307B','#5900FF']
# Title & Labels:
plt.title('Balanced Transaction Distribution by Hour', fontdict=FontT, loc='left')
plt.ylabel('Transaction Count'                       , fontdict=FontY)
plt.xlabel('Hour of the Day'                         , fontdict=FontX)
plt.xticks(range(0, 24))
plt.box(   False)
plt.show(       )
```
