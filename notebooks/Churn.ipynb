# Churn Prediction

Model predicting customer churn for TelCo.

Diving into TelCo sector analyzing customer churn, using Pandas **&** Machine Learning to study DataSets from TelCo firms, revealing demographic **&** usage patterns to predict customer retention, merging data analysis **&** predictive modeling with sharpen Data Science expertise.

The telecommunications **(**TeleCom**)** sector is rapidly changing, with more and more telecom businesses being created and many customers deciding to switch between providers. $Churn$ refers to the process where customers or subscribers stop using a company**'**s services or products. Understanding the factors that influence keeping a customer as a client in predicting churn is crucial for TeleCom companies to enhance their service quality **&** customer satisfaction.

*Churn rate*, or simply *churn*, represents the rate of evasion from the client base. For certain services, like SpotiFy ou NetFlix, it represents the rate of subscription cancellation, which is of extreme importance for the company, as an analysis over time may point there is some problem to be addressed.

From the dictionary definition, Churn Rate**:** percentage rate at which customers stop subscribing to a service or employees leave a job.

> *Churn é uma métrica que indica o quanto a empresa perdeu de receita ou clientes. Para calcular o churn, basta dividir a quantidade de clientes que perdeu até o final do período considerado pelo total de clientes que iniciaram. Por exemplo, se perdeu 10 clientes de 100, tem 10**%** de taxa de churn.*

    Churn      :          clientes desistentes
    Churn Rate = total de clientes cancelados / número total de clientes ativos no período considerado

*Churn* também pode ser usado para identificar potenciais cancelamentos, com alguma antecedência, promovendo ações direcionadas para tentar reter clientes, merecendo atenção, já que o Custo de Aquisição de Cliente **(**CAC**)** é normalmente mais alto que o custo para mantê**-**lo, não se desejando alto valor de *Churn Rate*.

> *Entender por que clientes abandonam o produto ou serviço é vital para conquistar crescimento sustentável. Como o churn tem um efeito negativo na receita de uma empresa, entender o que é esse indicador e como trabalhar para mitigar essa métrica é algo crítico para o sucesso de muitos negócios.*

Também pode ocorrer Early Churn quando o cancelamento ou a perda de cliente ocorre logo após a aquisição ou início do serviço, sem que tenha passado tempo significativo para gerado valor substancial para a empresa **(**valor do tempo de vida do cliente **–** LTV**:** LifeTime Value**)**, podendo indicar problemas com experiência do cliente, proposta de valor do produto, ou falhas de integração ou de suporte ao cliente, podendo vir ainda de vendas ruins, quando vendedores tentam empurrar produto a qualquer custo, mesmo que não seja ideal para o cliente **(**venda sob pressão**)**, sendo a taxa de churn métrica fundamental para entender a saúde do negócio, para orientar decisões estratégicas e para melhorar a experiência do cliente.

Churn Negativo ocorre quando a receita gerada por clientes existentes que permanecem com a empresa supera a perda de receita causada por clientes que saem.

Técnicas para melhorar churn **&** buscar churn negativo**:**

* Up**-**Selling **&** Cross**-**Selling**:** oferecer produtos ou serviços adicionais que aumentem o valor da conta de clientes existentes, levando a crescimento de receita geral, mesmo que alguns clientes estejam saindo;
* Expansão de Contas**:** aumento dos gastos de clientes existentes;
* Programa de Fidelidade ou de Recompensas**:** incentivo a lealdade de clientes, recompensando engajamento;
* Melhoria na Experiência do Cliente**:** excelente atendimento ao cliente **(**customer first**)** e melhorias contínuas em produto ou serviço;
* Segmentação **&** Personalização**:** oferecer soluções personalizadas que atendam melhor a necessidades de segmentos específicos de clientes.

No entanto, a melhor estratégia para redução de churn é fazer com que o produto ou serviço seja indispensável.

Ações para melhorar a taxa de churn, tornando**-**a mais aceitável**:**

* Create good first impression;
* Aligh expectations when closing a deal;

    InBound Marketing**:** oferecer conteúdo relevante e de valor para futuros clientes desde o começo do relacionamento, quando ainda não estão pensando em adquirir o produto ou serviço.

* Deliver the best experience;
* Listen to the client;
* Avoid the same mistakes, having an action plan.

    Focusing on clients success is the best way to go.

MRR Churn **(**Monthly Recurring ReVenue **–** Churn de Receita**):** receita perdida com o churn de clientes, detectando também DownGrades **&** UpGrades; sendo importante que os maiores e mais importantes clientes continuem ativos e crescendo, trazendo mais receita.

A taxa ideal de churn depende de muitos fatores, para S**aa**S **(**SoftWare as a Service**)**, por exemplo, é sugerido algo entre 5**%** a 7**%** ao ano. Modelo B2B **(**Business**-**to**-**Business**)** tende a ter a menor taxa de churn do mercado, reflexo de como são feitas contratações de serviços entre empresas, fechando contratos anuais e não sendo sensível a preços.

Customers switching from one company to another is called *churn*, and it is expensive all around**:** one company must spend on incentives to attract a customer while another company loses revenue when the customer departs. Marketing may designate a special retention offer.

All the analysis over clients data, revenues, and renovation are object of Big Data Analytics, using Machine Learning Models **–** automated algorithms that learns iteratively, helping to find InSights.

Dealing with ImBalanced Data in Churn Analysis

Datasets used for classification problem have a label **(**indicating the class**)** and a set of InPuts. Imbalanced data can have different ratios. There is no standard threshold for how your classes should be split.

[Metrics](https://towardsdatascience.com/accuracy-recall-precision-f-score-specificity-which-to-optimize-on-867d3f11124)**:**
* ⬆️Accuracy**:** total correct predictions over the total number of predictions.

    However, accuracy can be [misleading](https://machinelearningmastery.com/classification-accuracy-is-not-enough-more-performance-measures-you-can-use/) when working with ImBalanced DataSet.

    Because there are so many customers that do not churn, the high number of true negatives makes the accuracy very high, which is misleading since the focus is to discover what to do to prevent other customers from leaving.

* ⬆️Precision**:** out of all customers labeled as churn, how many were correctly labeled as such, that is, the ratio of customers correctly labeled **(**True Positives**)** to all customers labeled as churn **(**True Positives **&** False Positives**)**.
* ⬆️[AUC](https://towardsdatascience.com/understanding-auc-roc-curve-68b2303cc9c5)**:** Area Under the Curve **&** [ROC](https://www.youtube.com/watch?v=4jRBRDbJemM)**:** Receiver Operator Characteristic.
* ⬆️F1**-**Score**:**  commonly used in information retrieval, measures accuracy using precision **&** recall, weighting both equally. Hence, good retrieval algorithm would maximize both precision **&** recall simultaneously. Thus, moderately good performance on both will be favored over extremely good performance on one **&** poor performance on the other.
    - Precision**:** the ratio of true positives **(**TP**)** to all predicted positives **(**TP **+** FP**)**;
    - ReCall**:** the ratio of true positives **(**TP**)** to all actual positives **(**TP **+** FN**)**.

* ⬇️MAE   **$-$**      Mean Absolute            Error **(**Média Absoluta do Erro**)**;
* ⬇️MSE   **$-$**      Mean Squared             Error **(**Erro Quadrático Médio**)**;
* ⬇️RMSE  **$-$** Root Mean Squared             Error **(**Raiz do Erro Quadrático Médio**)**;
* ⬇️RMSLE **$-$** Root Mean Squared  Log        Error;
* ⬇️MAPE  **$-$**      Mean Absolute Percentage Error **(**Média Percentual do Erro Absoluto**)**;
* ⬆️R2 **(**R2 Score**)** **$-$** Coefficient of Determination.

Balancing Data Techniques**:**
* UnderSampling**:** deleting observations from the over**-**represented class, meaning active customers; making sense when there are a whole LOT of data **("**tens**-** or hundreds of thousands of instances or more**")**.
* OverSampling**:** adding copies of observations from the under**-**represented class, being an OK approach as it balances the data, but there are better options.
* SMOTE**:** Synthetic Minority Over**-**Sampling Technique **–** creating synthetic samples of the data, using distance measure to create synthetic samples of data points that would not be far from the original data points.
* ADASYN**:** Adaptive Synthetic Sampling Method **–** an improved way to apply SMOTE, looking at the density distribution of the data to determine the number of samples that will be generated.

```python
# Converting to Float:
# df['total']=df['total'].astype(float)
def strTOfloat(data):
    '''
    Convert string to float.
    # Arguments:
           data: string.
    # Returns:
        Vale converted to float, if conversion is possible;
        NaN, in case value is not appropriate.
    '''
    try: return float(data)
    except ValueError: return np.nan
# Copy DataFrame:
clean=DF.copy()
# Remove ID Column: not relevant for analysis
clean.drop('id', axis=1, inplace=True)
# Convert TotalCharges to Float:
clean['total']=clean['total'].apply(strTOfloat)
# Verifying Missing Entries:
print("Missing Entries in 'total' Replaced with Median: {}.\n".format(clean['total'].isnull().sum()))
# clean.loc[clean['total'].isnull()]
# RePlacing Missing Entries with Median: no effect on data
TotalChargesMedian=clean['total'].median()
clean['total'].fillna(TotalChargesMedian, inplace=True)
print('Variable Types:\n{}'.format(clean.dtypes))
```

```python
# @title BaseLines
# Validation:
def ModelValidation(X, Y, clf, quite=False):
    '''
    Realiza cross-validation com os dados de treino para determinado modelo.
    # Arguments:
        x      : DataFrame, contém as variáveis independetes.
        y      : Series, vetor contendo a variável alvo.
        clf    : modelo classificador do Scikit-learn.
        quite  : bool, indicando se a função deve imprimir os resultados ou não.
    # Returns  :
        float  , média dos scores da cross-validation.
    '''
    X          = np.array(X)
    Y          = np.array(Y)
    pipeline   = make_pipeline(StandardScaler(), clf)
    scores     = cross_val_score(pipeline, X, Y, scoring='recall')
    if  quite == False: print('{}\n      ReCall\t {:.2f}% +/- {:.0f}%'.format(clf, scores.mean()*100, scores.std()*100))
    return scores.mean(       )
# BaseLine Models:
lr   =LogisticRegression(     )
rd   =RidgeClassifier(        )
dt   =DecisionTreeClassifier( )
rf   =RandomForestClassifier( )
nb   =GaussianNB(             )
knn  =KNeighborsClassifier(   )
svm  =SVC(                    )# Support    Vector            Machine
gbm  =LGBMClassifier(         )# Light      Gradient-Boosting Machine   https://lightgbm.readthedocs.io/en/latest/pythonapi/lightgbm.LGBMClassifier.html
sgd  =SGDClassifier(          )# Stochastic Gradient Descent
xgb  =XGBClassifier(          )# eXtreme    Gradient Boosting           https://xgboost.readthedocs.io/en/stable/index.html
# ImBalanced BaseLine:
model=[                       ]
score=[                       ]
for clf in  (lr, rd, dt, rf, nb, knn, svm, gbm, sgd, xgb):
    model.append(clf.__class__.__name__)
    score.append(ModelValidation(trainX, trainY, clf, quite= True ))
pd.DataFrame(data=score,   index=model ,         columns=['ReCall'])
```

```python
# @title PipeLines
lrPL  =Pipeline([('lr' , LogisticRegression(    ))])
rdPL  =Pipeline([('rd' , RidgeClassifier(       ))])
dtPL  =Pipeline([('dt' , DecisionTreeClassifier())])
rfPL  =Pipeline([('rf' , RandomForestClassifier())])
nbPL  =Pipeline([('nb' , GaussianNB(            ))])
knnPL =Pipeline([('knn', KNeighborsClassifier(  ))])
svmPL =Pipeline([('svm', SVC(                   ))])
gbmPL =Pipeline([('gbm', LGBMClassifier(        ))])
sgdPL =Pipeline([('sgd', SGDClassifier(         ))])
xgbPL =Pipeline([('xgb', XGBClassifier(         ))])
# Dictionary:
dictPL={0:        'lr ',
        1:        'rd ',
        2:        'dt ',
        3:        'rf ',
        4:        'nb ',
        5:        'knn',
        6:        'svm',
        7:        'gbm',
        8:        'sgd',
        9:        'xgb'}
X        =trainX
Y        =trainY
# PipeLines Fitting:
pipelines=[lrPL, rdPL, dtPL, rfPL, nbPL, knnPL, svmPL, gbmPL, sgdPL, xgbPL]
for pipe in pipelines: pipe.fit(X, Y)
cvResults=[                         ]
for i, model in  enumerate(pipelines):
    cvScore=cross_val_score(model, X, Y,  groups=None, scoring='recall', cv=kFold, n_jobs=-1, verbose=0,
                            params=None, pre_dispatch='2*n_jobs',     error_score=np.nan )
    cvResults.append(cvScore)
    print('%s: %.2f ' % (dictPL[i], cvScore.mean()))
```
