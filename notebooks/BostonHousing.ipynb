# [Boston Housing](https://kaggle.com/c/boston-housing)

<center>
    <img src=https://images.unsplash.com/photo-1603298108410-e6f28ad2708d>
</center>

Boston Housing DataSet**:** from information collected by the _U**.**S**.** Census Service_ in Boston**/**MA.

Objective**:** predict real estate **(**selling**)** prices.

Variables**:**

* `crim`**:** per capita crime rate by town;
* `zn`**:** proportion of residential land zoned for lots over 25**,**000 sq**.**ft**.** **(~**2300m$^2$);
* `indus`**:** proportion of non**-**retail business acres per town;
* `chas`**:** Charles River dummy variable **(**1 if tract bounds river; 0 otherwise**)**;
* `nox`**:** nitrogen oxides concentration **(**parts per 10 milhões**)**;
* `rm`**:** average number of rooms per dwelling;
* `age`**:** proportion of owner**-**occupied units built prior to 1940;
* `dis`**:** weighted mean of distances to five Boston employment centers;
* `rad`**:** index of accessibility to radial highways;
* `tax`**:** full**-**value property**-**tax rate per 10**,**000;
* `ptratio`**:** pupil**-**teacher ratio by town;
* `black`**:** 1000 **(**BK **–** 0**.**63**)^**2 where BK is the proportion of blacks by town;
* `lstat`**:** lower status of the population **(**percent**)**;
* `mdev`**:** median value of owner**-**occupied homes in 1**,**000.

Target Variable**:** `mdev`; giving its continuous nature, it is a regression problem.

DataSet Split In order to avoid Data Leaking.

`setup()`**:** initiate the PyCaret environment, creating a transformation PipeLine to prepare data for modeling **&** deploy, and it must be executed prior to any other PyCaret function, with two _obrigatory parameters_**:**
* Pandas DataFrame;
* Target Variable.

Other parameters are optional, used to customize the preprocessing PipeLine.

`setup()`**:** PyCaret infers automatically the type of each variable based on certain properties, displaying a table with features and its infered types.

It is fundamental that data types be correct because it automatically executes some preprocessing needed for Machine Learning problems, which are different for each type of data, that is why it is essential they are correctly configured.

Once correctly executed, SetUp prints a grid with important information about the preprocessing PipeLine**:**
* Original Data**:** original dataset size;
* Missing Values**:** `True` when there is missing data on the DataSet;
* Numeric Features**:** number of features infered as numeric;
* Categorical Features**:** number of features infered as categorical;
* Transformed Train Set**:** train set shape. Original is **(**455**,** 14**)**, but it has been transformed to **(**328**,** 23**)** for the train dataset. Features has grown from 14 to 23 due to categorical features encoding;
* Transformed Test Set**:** test**/**validation shape. Split done using default `train_size` setting **(**70**/**30**)**.

Some essential tasks for modeling are automatically done, like dealing with missing values, encoding categorical variables, etc. Most of parameters **&** possibilities can be customized in order to have more control over those settings.

## Comparing Models

`compare_models(sort='R2', fold=10)`

Treina todos os modelos na biblioteca do PyCaret e os avalia usando validação cruzada e retornando tabela com a média de MAE, MSE, RMSE, R2, RMSLE e MAPE dos folds **(**10, por padrão**)** de todos os modelos disponíveis na biblioteca.
* ⬇️MAE   **$-$**      Mean Absolute            Error **(**Média Absoluta do Erro**)**;
* ⬇️MSE   **$-$**      Mean Squared             Error **(**Erro Quadrático Médio**)**;
* ⬇️RMSE  **$-$** Root Mean Squared             Error **(**Raiz do Erro Quadrático Médio**)**;
* ⬆️R2 **(**R2 Score**)**    **$-$** Coefficient of Determination;
* ⬇️RMSLE **$-$** Root Mean Squared  Log        Error;
* ⬇️MAPE  **$-$**      Mean Absolute Percentage Error **(**Média Percentual do Erro Absoluto**)**.

```python
# @title Settings
# Libraries:
import      numpy                   as   np
import     pandas                   as   pd
import     plotly.express           as   px
import matplotlib.pyplot            as   plt
import    seaborn                   as   sns
import                                   sys
from      pycaret.datasets        import get_data
from      pycaret.regression      import    *
# Configurations:
print('System Ready: Python –',          sys.version)
%matplotlib inline
%config     InlineBackend.figure_format='svg'
%precision 2
pd.options.plotting.matplotlib.register_converters=True
pd.set_option('display.max.columns',               None)
plt.rcParams[      'font.family']      =                                        'sans-serif'
sns.set_theme(context='notebook', style='whitegrid', palette='colorblind', font='sans-serif', font_scale=1.15, color_codes=True, rc={'grid.color':'1','grid.linestyle':':'})
```

```python
# @title Data
# df=pd.read_csv('https://raw.githubusercontent.com/rafaelnduarte/sigmoidal_data/master/HousingData.csv')
data=get_data('boston')
data
```

```python
# @title Scatter Plots
fig=px.scatter(x=data['lstat'], y=data['medv'], facet_col=data['rad'], opacity=.2,
              trendline='ols' ,  trendline_color_override= '#FF103F', labels={'y':'','x':''})
fig.show(     )
```

```python
# @title HistoGram
fig=px.histogram(data, x=['medv'], labels={'count':'','value':''})
fig.show(       )
```

```python
# @title Train & Test Split
TE=data.sample(frac=.10)# test
TA=data.drop(  TE.index)# train
TE.reset_index(inplace=True, drop=True)
TA.reset_index(inplace=True, drop=True)
print(TA.shape,'\n',TE.shape)
```

```python
# @title Model SetUp
regression=setup(data=TA, target='medv', train_size=.7, fold=5, session_id=0, experiment_name='BostonHousing')
```

```python
# @title Model Comparisson
best=compare_models(fold=5, round=2, sort='RMSE', n_select=1)
```

```python
# @title Best Model ParaMeters
best.get_params()
```

```python
# @title Model Tuning
tune_model(best, round=2, optimize='RMSE')
```

```python
# @title AutoML
automl(optimize='RMSE', return_train_score=True)
```

```python
# @title Model Evaluation
evaluate_model(best)
```

```python
# @title Prediction Error
plot_model(best, plot='error', save=False, display_format=None)# 'streamlit'
```

```python
# @title Features Importance
plot_model(best, plot='feature')# feature_all
```

```python
# Model Finalization:
model=finalize_model(best)
# @title Model Saving
save_model(model,'BostonHousingModel')
```

```python
# @title Model Loading
model=load_model('BostonHousingModel')
```

```python
# @title Model Prediction
predict_model(model, data=TE, round=2, verbose=True)
```
