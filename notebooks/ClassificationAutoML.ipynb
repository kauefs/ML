Classification**:** SuperVised Machine Learning Module used for classifying elements into groups to predict categorical class labels discrete **&** unordered.

Objective**:** Fetal Health State to Prevent Child **&** Maternal Mortality.

Reduction of child mortality is reflected in several of the United Nations**'** Sustainable Development Goals and is a key indicator of human progress. The UN expects that by 2030, countries end preventable deaths of newborns and to children under 5 years of age, with all countries aiming to reduce under**-**5 mortality to at least as low as 25 per 1**,**000 live births. Parallel to notion of child mortality is of course maternal mortality, which accounts for 295**,**000 deaths during **&** following pregnancy **&** childbirth **(**as of 2017**)**. The vast majority of these deaths **(**94**%)** occurred in low**-**resource settings, and most could have been prevented.

Cardiotocography **(**CTG**)** is used during pregnancy to monitor fetal heart rate **&** uterine contractions, monitoring fetal well**-**being, allowing for early detection of fetal distress, as CTG interpretation helps determining if a pregnancy is of high or of low risk, with abnormal CTG indicating need for further investigations **&** potential intervention.

CTG are simple **&** cost accessible option to assess fetal health, allowing healthcare professionals to take action in order to prevent child **&** maternal mortality. The equipment itself works by sending ultrasound pulses and reading its response, thus shedding light on fetal heart rate **(**FHR**)**, fetal movements, uterine contractions and more.

Machine Learning PipeLine to predict fetal health based on features derived from cardiotocogram **(**CTG**)** readings to classify fetal health into three categories**:**
1. Normal;
2. Suspect;
3. Pathological.

Prolongued Decelerations**:** an abnormal fetal heart rate **(**FHR**)** deceleration that usually lasts 2 to 10 minutes but may last longer.
- Lower Number**:** lower mean value for prolonged decelerations is generally better, suggesting prolonged decelerations are less frequent, typically indicating fewer instances of potential fetal distress and a healthier pregnancy;
- Higher Number**:** higher mean value would indicate more frequent prolonged decelerations, which could be a sign of increased fetal distress or complications, usually warranting closer monitoring and possibly medical intervention to ensure fetal well**-**being.

Abnormal Short Term Variability**:** percentage of time the fetal heart rate exhibits abnormal short**-**term variability.
- Normal Values**:** typically, the percentage of time with abnormal short**-**term variability should be low in a healthy pregnancy. A value close to 0**%** would be considered normal;
- Abnormal Values**:** higher percentages indicate a significant portion of the monitoring period showed abnormal short**-**term variability.
    - Moderately Abnormal**:** values ranging from 20**%** to 50**%**;
    -   Severely Abnormal**:** values above 50**%**, indicating more than half of the monitoring period exhibited abnormal variability.

Early prediction of fetal health conditions can assist healthcare professionals in taking preventive measures, thus reducing risks to both the fetus and the mother. It is common that when things go wrong, it is not just one of them but a combination of them**:**
* ProLongued Decelerations **+** Severe Decelerations **+** Abnormal Short Term Variability;
* Abnormal Short Term Variability **+** PerCentage of Time with Abnormal Long Term Variability;
* ProLongued Decelerations **+** Abnormal Short Term Variability **+** BaseLine Value Out of Ordinary.

Naive Bayes**:** for non complex data **&** relatively simple task **–** a high**-**bias**/**low**-**variance classifier, with advantages over logistic regression **&** nearest neighbor algorithms when working with limited amount of data available to train a model. It is also a good choice when CPU **&** memory resources are a limiting factor, and because it is very simple, it does not tend to overfit data, and it can be trained very quickly, doing well with continuous new data used to update the classifier.

$k$Nearest Neighbors **(**$k$NN**):** categorize data points based on their distance to other points in a training data set **–** an instance**-**based lazy learner, which means there is no real training phase. The training data is loaded into the model and let sit until starting to use the classifier. With a new query instance, the KNN model looks for the specified $k$ number of nearest neighbors; if $k$ is 5, it finds the class of five nearest neighbors. In order to apply label**/**class, the model takes a vote to see where it should be classed.

Decision Trees**:** to see how a decision tree predicts a response, follow the decisions in the tree from the root node down to a leaf node which contains the response. Classification trees give responses that are nominal, such as true or false. Regression trees give numeric responses. They are also relatively quick.
